## 7. Filter rows

(a) Two or more bedrooms

```python
mask = listings['bedrooms'] >= 2

print(mask.sum())
print(mask.mean())
```

1,825 listings, about 39.9% of all listings.

(b) Entire homes in Östermalms with two or more bedrooms

Intentional error using `and`

```python
selected = listings[
    (listings['neighbourhood_cleansed'] == 'Östermalms')
    & (listings['room_type'] == 'Entire home/apt')
    & (listings['bedrooms'] >= 2)
]

print(selected.shape)
```

```python
selected = listings[
    (listings['neighbourhood_cleansed'] == 'Östermalms')
    & (listings['room_type'] == 'Entire home/apt')
    & (listings['bedrooms'] >= 2)
]

print(selected.shape)
```
